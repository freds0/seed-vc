## Step 3 Pre-training evaluation

Let's see how does our model perform under zero-shot inference, that is, without training on these speakers

```python
%cd /content/seed-vc/
!python inference.py --source examples/source/source_s1.wav \
--target ../dataset/Taffy/Taffy_1.wav \
--output ./reconstructed \
--diffusion-steps 25 \
--length-adjust 1.0 \
--inference-cfg-rate 0.7

!python inference.py --source examples/source/source_s1.wav \
--target ../dataset/Kobe/Kobe_1.wav \
--output ./reconstructed \
--diffusion-steps 25 \
--length-adjust 1.0 \
--inference-cfg-rate 0.7

import librosa
sr = 22050
source_wave, _ = librosa.load("examples/source/source_s1.wav", sr=sr)
ref_wave_1, _ = librosa.load("../dataset/Taffy/Taffy_1.wav", sr=sr)
ref_wave_2, _ = librosa.load("../dataset/Kobe/Kobe_1.wav", sr=sr)
vc_wave_1 = librosa.load("./reconstructed/vc_source_s1_Taffy_1_1.0_25_0.7.wav", sr=sr)[0]
vc_wave_2 = librosa.load("./reconstructed/vc_source_s1_Kobe_1_1.0_25_0.7.wav", sr=sr)[0]

# Display with IPython, with respective names
from IPython.display import Audio, display
print("Source speech")
display(Audio(source_wave, rate=sr))
print("Reference speech Taffy")
display(Audio(ref_wave_1, rate=sr))
print("Converted speech Taffy")
display(Audio(vc_wave_1, rate=sr))
print("Reference speech Kobe")
display(Audio(ref_wave_2, rate=sr))
print("Converted speech Kobe")
display(Audio(vc_wave_2, rate=sr))
```

## Step 4 Launch Training

Similarity is not bad, but we hope to make it better since we have training script and a lot of data for each speaker  

Let's train for 200 steps with batch size of 4, it will take about 10 minutes

```python
!python train.py --config configs/presets/config_dit_mel_seed_uvit_whisper_small_wavenet.yml \
--dataset-dir ../dataset \
--run-name Kobe_Taffy_ft \
--batch-size 4 \
--max-steps 200 \
--max-epochs 1000 \
--save-every 100 \
--num-workers 2
# Both checkpoint and results will be saved to ./runs/<run-name>/
```

## Step 5 Post-training Evaluation

```python
# You can definitely train for more steps, but we end this for now as it is for demo only.
# lets try inference with the fine-tuned checkpoint
# Let's see how does our model perform under zero-shot inference, that is, without training on these speakers
%cd /content/seed-vc/
!python inference.py --source examples/source/source_s1.wav \
--target ../dataset/Taffy/Taffy_1.wav \
--output ./reconstructed \
--diffusion-steps 25 \
--length-adjust 1.0 \
--inference-cfg-rate 0.7 \
--config-path ./runs/Kobe_Taffy_ft/config_dit_mel_seed_uvit_whisper_small_wavenet.yml \
--checkpoint-path ./runs/Kobe_Taffy_ft/ft_model.pth

!python inference.py --source examples/source/source_s1.wav \
--target ../dataset/Kobe/Kobe_1.wav \
--output ./reconstructed \
--diffusion-steps 25 \
--length-adjust 1.0 \
--inference-cfg-rate 0.7 \
--config-path ./runs/Kobe_Taffy_ft/config_dit_mel_seed_uvit_whisper_small_wavenet.yml \
--checkpoint-path ./runs/Kobe_Taffy_ft/ft_model.pth

import librosa
sr = 22050
source_wave, _ = librosa.load("examples/source/source_s1.wav", sr=sr)
ref_wave_1, _ = librosa.load("../dataset/Taffy/Taffy_1.wav", sr=sr)
ref_wave_2, _ = librosa.load("../dataset/Kobe/Kobe_1.wav", sr=sr)
vc_wave_1 = librosa.load("./reconstructed/vc_source_s1_Taffy_1_1.0_25_0.7.wav", sr=sr)[0]
vc_wave_2 = librosa.load("./reconstructed/vc_source_s1_Kobe_1_1.0_25_0.7.wav", sr=sr)[0]

# Display with IPython, with respective names
from IPython.display import Audio, display
print("Source speech")
display(Audio(source_wave, rate=sr))
print("Reference speech Taffy")
display(Audio(ref_wave_1, rate=sr))
print("Converted speech Taffy")
display(Audio(vc_wave_1, rate=sr))
print("Reference speech Kobe")
display(Audio(ref_wave_2, rate=sr))
print("Converted speech Kobe")
display(Audio(vc_wave_2, rate=sr))
```

Sounds much better! Of course more training steps on more data will be helpful, which you can try out by yourself.
